# training process

###Importing required libraries for splitting data and model creation....


```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

#----Defining Features (X) and Targets (Y)
#---X = input fetures for prediction
#---Y = targets variable which we want to predict ( House price)

```python
X = df[['Area_sqft','Bedrooms','Bathrooms','Age_of_House',]]
y = df['Price']
```

```python
X_train, X_test, y_train, y_test = train_test_split ( X,y , test_size =0.2)
print(X_train, X_test, y_train, y_test)
```

### creating an instance of Linear Regression model..
#---This is base line model to pradict house price from features like area , bedroom ,etc
#---after printing the model object to verify its initialization with default parrameters....

```python
model = LinearRegression() 
print(model)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)### prediction on test data......
print(y_pred)
```

```python
from sklearn.metrics import mean_squared_error,r2_score ,mean_absolute_error
```

```python
###-----Evaluating model performance-------..
print(mean_squared_error(y_test, y_pred))     ###----Calculates MSE-----#
print(r2_score(y_test, y_pred)*100)           ###----R2 Score - how well model explains variance
print(mean_absolute_error(y_test, y_pred))    ###----Calculates MAE---
```

## Ridge Regression
it uses l2 regularization...

```python
from sklearn.linear_model import Ridge
Ridge_model = Ridge(alpha=0.5)                ##----Initializing Ridge model with alpha = 0.5(regularization strength) 
Ridge_model.fit(X_train, y_train)             ##----Training the model on training data-----
y_pred = Ridge_model.predict(X_test)          ###----prediction on test data-----
print(mean_squared_error(y_test, y_pred))     ##---Calulates MSE----
print(r2_score(y_test, y_pred)*100)           ## R2 Score - how well model explains variance-------
print(mean_absolute_error(y_test, y_pred))    ##---Calculates MAE------
```

## LASSO..

```python
from sklearn.linear_model import Lasso      ###-----L1 regularization helps in feature select by making some coefficients zero 
Lasso_model = Lasso(alpha=0.5)         ####----Initializing Lasso model with alpha == #####0.5
Lasso_model.fit(X_train, y_train)        ###----Training the model.... 
y_pred = Lasso_model.predict(X_test)     ###----predicting on test data.....
print(mean_squared_error(y_test, y_pred))  ####----Calculation MAE.......
print(r2_score(y_test, y_pred)*100)        ###----how well model explains variance......
```

```python
### ---- CHART AFTER LASSO MODEL ---- 
import matplotlib.pyplot as plt
```

```python
##----- Actual vs predicted chart  for lasso ----
plt.figure(figsize=(6,4))                  #-----set chart size-----#
plt.scatter(y_test,y_pred, alpha=0.6 , color = 'green')     ##------plot actual vs predicted points-----#
plt.xlabel("Actual Price")       ###-----label x axis-----#
plt.ylabel("Predited Price")     ###-----label for y axis-----#
plt.title("LASSO -  Actual  vs Prediction")## chart title----#
## prediction line
plt.plot([y_test.min(), y_test.max()],[y_test.min(), y_test.max()], 'r--')    ### red dotted line for perfect predicton
plt.show() ###--- display the chart---#
```
